# Key Insights from Superstore Sales Analysis

## 1. Dataset Overview
- The dataset contains **9,994 orders** with **21 columns**, no null values, and no duplicate rows — data is clean.
- Data spans **January 2014 to December 2017** (4 years).
- Sales are right-skewed: mean is **229.86**, median is **54.49**, and max is **22,638** — most orders are small but a few are very large.

## 2. Sales by Region
- **West** is the top-performing region with **725,458** in total sales.
- **East** is second at **678,781**.
- **South** is the weakest at **391,721** — roughly half of West.
- Central and South regions have clear room for growth.

## 3. Profit by Region
- **West** leads profit at **108,418**.
- **Central** has the lowest profit at **39,706** despite being 3rd in sales — suggesting heavy discounting or high product costs in that region.

## 4. Profit by Category
- **Technology** is the most profitable category at **145,455**.
- **Office Supplies** is second at **122,491**.
- **Furniture** generates only **18,451** in profit — very poor margins despite its sales volume.

## 5. Monthly Sales Trend
- Sales follow a **clear seasonal pattern** — peaks occur in **Q4 (Nov-Dec)** every year.
- There is a strong **upward trend** from 2014 to 2017, with the highest single month (~118,000) in **November 2017**.
- Sales consistently dip in **Jan-Feb** and recover from mid-year.

## 6. Top 10 Most Profitable Products
- **Canon imageCLASS 2200 Advanced Copier** is the single most profitable product.
- Other top products: **Fellowes PB500 Binding Machine**, **HP LaserJet 3310 Copier**, **Canon PC1060 Laser Copier**.
- High-end **Copiers and Printers** dominate the top 10 profit-makers.

## 7. Bottom 10 Loss-Making Products
- **Cubify CubeX 3D Printers** (Double and Triple Head) are the biggest loss-makers at nearly -8,880 and -3,840.
- **Lexmark Laser Printer** and **Chromcraft Conference Tables** also cause major losses.
- Most loss-makers belong to **Furniture (Tables)** and **Technology (3D Printers)**.

## 8. Profit by Sub-Category
- Top 3 profit sub-categories: **Copiers** (55,617), **Phones** (44,516), **Accessories** (41,936).
- Loss-making sub-categories: **Tables** (-17,725), **Bookcases** (-3,473), **Supplies** (-1,189).
- Furniture sub-categories consistently lose money and need a pricing strategy review.

## 9. Discount Impact on Profit
- Discounts of **0-20%** are associated with **positive average profit**.
- Discounts above **30%** lead to **negative average profit** consistently.
- At **50% discount**, the average profit drops to **-310 per order**.
- **Business Recommendation: Never exceed 20% discount** — higher discounts directly destroy profitability.

# Recommendation 1 — Fix the Furniture Problem

Furniture category is generating the lowest profit, primarily driven by Tables (loss of ₹17K) and Bookcases. The business should consider repricing these products or discontinuing heavy discounts on them. Eliminating discounts on Tables alone could recover significant losses.

# Recommendation 2 — Control Discounting Strategy

Data clearly shows higher discounts directly cause losses. Business should cap discounts at 20% maximum across all categories. This single change could significantly improve overall profit margins.

# Recommendation 3 — Capitalize on Year End Sales Spike

November and December consistently show highest sales across 2015, 2016 and 2017. Business should increase inventory and marketing budget in October every year to maximize this seasonal opportunity.

# SALES by REGION

```python
import matplotlib.pyplot as plt
import seaborn as sns

region_sales = (
    df.groupby('Region')['Sales']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

sns.barplot(
    x=region_sales.index,
    y=region_sales.values,
    hue=region_sales.index,
    palette='Blues_d',
    legend=False
)

plt.title('Total Sales by Region', fontsize=16)
plt.xlabel('Region')
plt.ylabel('Total Sales ($)')

plt.tight_layout()

plt.savefig('sales_by_region.png')

plt.show()
```

# PROFIT by CATEGORY

```python
import matplotlib.pyplot as plt
import seaborn as sns

cat_profit = (
    df.groupby('Category')['Profit']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,6))

colors = [
    'green' if x > 0 else 'red'
    for x in cat_profit.values
]

sns.barplot(
    x=cat_profit.index,
    y=cat_profit.values,
    hue=cat_profit.index,
    palette=colors,
    legend=False
)

plt.title('Total Profit by Category', fontsize=16)

plt.xlabel('Category')
plt.ylabel('Total Profit ($)')

plt.tight_layout()

plt.savefig('profit_by_category.png')

plt.show()
```

# SUB CATEGORY LOSS

```python
import matplotlib.pyplot as plt
import seaborn as sns

subcat_profit = (
    df.groupby('Sub-Category')['Profit']
      .sum()
      .sort_values(ascending=True)
)

plt.figure(figsize=(12,8))

colors = [
    'red' if x < 0 else 'green'
    for x in subcat_profit.values
]

sns.barplot(
    x=subcat_profit.values,
    y=subcat_profit.index,
    hue=subcat_profit.index,
    palette=colors,
    legend=False
)

plt.title(
    'Profit by Sub-Category (Red = Loss)',
    fontsize=16
)

plt.xlabel('Total Profit ($)')

plt.tight_layout()

plt.savefig('profit_by_subcategory.png')

plt.show()
```

# MONTHLY TREND

```python
import matplotlib.pyplot as plt

df['Order Month'] = (
    df['Order Date']
      .dt.to_period('M')
)

monthly_sales = (
    df.groupby('Order Month')['Sales']
      .sum()
)

monthly_sales.index = (
    monthly_sales.index.astype(str)
)

plt.figure(figsize=(14,6))

monthly_sales.plot(
    kind='line',
    color='blue',
    linewidth=2
)

plt.title(
    'Monthly Sales Trend (2014–2017)',
    fontsize=16
)

plt.xlabel('Month')
plt.ylabel('Total Sales ($)')

plt.tight_layout()

plt.savefig('monthly_sales_trend.png')

plt.show()
```
